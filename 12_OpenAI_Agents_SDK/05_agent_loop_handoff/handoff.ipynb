```python
import os
from dotenv import load_dotenv
load_dotenv()

secret = os.environ.get('GEMINI_API_KEY')
MODEL="gemini/gemini-2.0-flash"
```

```python
import asyncio
from agents import Agent, Runner,function_tool
from agents.extensions.models.litellm_model import LitellmModel

@function_tool
def weather_tool(city):
    if city == 'Lahore':
        return f'The weather in {city} is 33deg Celsius'
    if city == 'Faislabad':
        return f'The weather in {city} is 45deg Celsius'
    
    return f'The weather in {city} is 20deg Celsius'



weather_agent = Agent(
    name="Weather Agent",
    instructions="You are a specialized weather agent",
    model=LitellmModel(model=MODEL),
    tools=[weather_tool],
    handoff_description="You are a specialized weather agent. You will answer all queries related to the weather",
)


general_agent = Agent(
    name="General_Assistant",
    instructions="You will answer general questions, and handoffs specialized topics",
    model=LitellmModel(model=MODEL),
    handoffs=[weather_agent]
)


# result1 = await Runner.run(general_agent,'what is recurrsion in haiku style')
result2 = await Runner.run(general_agent,'What is the weather of Faislabad')
result3 = await Runner.run(general_agent,'What is the weather of lahore')
result4 = await Runner.run(general_agent,'What is the weather of LA')
# print(result1.final_output)
print(result2.final_output)
print(result3.final_output)
print(result4.final_output)
```
