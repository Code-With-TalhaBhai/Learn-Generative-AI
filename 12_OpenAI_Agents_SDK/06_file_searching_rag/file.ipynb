#### User Given File Searching RAG

```python
from agents import Runner, FileSearchTool, Agent

agent = Agent(
    name="file_searching_agent",
    tools=[
        FileSearchTool(
            max_num_results=3,
            vector_store_ids=['vs_69394a7b06788191a777e0558514a46c']
        ),
    ],
    model='gpt-4.1-mini'
)

result = await Runner.run(agent, 'Who is Talha')
print(result.final_output)
```

```python
result2 = await Runner.run(agent,'What technologies he is working on? Is it in-demand or outdated technologies')
print(result2.final_output)
```

```python
result3 = await Runner.run(agent, 'What do you do suggest to him, what would he do to become better full stack developer or valuable in the market, any new skill or improve full stack developer skill')
print(result3.final_output)
```
